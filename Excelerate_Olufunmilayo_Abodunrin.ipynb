**Excercise 3**: Lists

```python
monthly_expenses= [2500, 3000, 2800, 3200, 2900]
total_expenses = sum(monthly_expenses)
average_expenses = total_expenses / len(monthly_expenses)

minimum = min(monthly_expenses)
maximum = max(monthly_expenses)

print(f"Total Expenses: ${total_expenses}")
print(f"Average Expenses: ${average_expenses}")
print(f"Minimum Expense: ${minimum}")
print(f"Maximum Expense: ${maximum}")

monthly_expenses.append(3100)

new_total_expenses = sum(monthly_expenses)
new_average_expense = new_total_expenses / len(monthly_expenses)

print(f"New Total Expenses: ${new_total_expenses}")
print(f"New Average Expense: ${new_average_expense:.2f}")


```

**Exercise 8**: NumPy Array Operations


```python
import numpy as np

product_a_sales = np.array([120, 150, 130, 180, 160])
product_b_sales = np.array([100, 130, 140, 170, 150])

total_sales_per_month = product_a_sales + product_b_sales
print("Total sales per month:", total_sales_per_month)

difference_in_sales = product_a_sales - product_b_sales
print("Difference in sales (A - B):", difference_in_sales)

average_sales_a = np.mean(product_a_sales)
average_sales_b = np.mean(product_b_sales)
print(f"Average sales for Product A: {average_sales_a}")
print(f"Average sales for Product B: {average_sales_b}")

is_a_greater = product_a_sales > product_b_sales
print("Is Product A greater each month?:", is_a_greater)

count_a_greater = np.sum(is_a_greater)
print("Number of months Product A had higher sales:", count_a_greater)
```

**Exercise 9**:  Matplotlib Basic Plotting


```python
import matplotlib.pyplot as plt

# Line Plot
x_values = [1, 2, 3, 4, 5]
y_values = [2, 4, 5, 4, 5]

plt.figure(figsize=(7, 4))
plt.plot(
    x_values,
    y_values,
    color='teal',          # unique color
    linestyle='--',
    marker='o',            # circle markers
    markerfacecolor='orange',
    linewidth=2,
    label='Data Series'
)
plt.title("Stylish Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()

# Bar Chart
categories = ['A', 'B', 'C']
values = [10, 15, 7]
bar_colors = ['#FF6F61', '#6B5B95', '#88B04B']

plt.figure(figsize=(6, 4))
plt.bar(categories, values, color=bar_colors, edgecolor='black', linewidth=1.2)
plt.title("Custom-Colored Bar Chart")
plt.xlabel("Category")
plt.ylabel("Values")
plt.show()
```

**Exercise 10:** Seaborn Basic Plotting

```python
# --- Import Libraries ---
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# --- Create Sample Data ---
category_data = pd.DataFrame({
    'Category': ['Type A', 'Type B', 'Type A', 'Type C', 'Type B', 'Type C'],
    'Value': [10, 15, 12, 8, 18, 10]
})

# --- Set Modern Style ---
sns.set_theme(style="whitegrid")

# --- 1. Count Plot ---
plt.figure(figsize=(6, 4))
sns.countplot(
    x='Category',
    data=category_data,
    palette='crest',
)
plt.title('Count of Each Category', fontsize=14)
plt.xlabel('Category')
plt.ylabel('Count')
plt.grid(axis='y', linestyle=':', alpha=0.5)
plt.show()

# --- 2. Bar Plot (Average Value by Category) ---
plt.figure(figsize=(6, 4))
sns.barplot(
    x='Category',
    y='Value',
    data=category_data,
    hue='Category',
    palette='flare',
    errorbar=None,
    legend=False
)
plt.title('Average Value per Category', fontsize=14)
plt.xlabel('Category')
plt.ylabel('Average Value')
plt.grid(axis='y', linestyle=':', alpha=0.5)
plt.show()

# --- 3. Correlation Heatmap ---
corr_matrix = np.array([
    [1.00, 0.75, 0.60],
    [0.75, 1.00, 0.55],
    [0.60, 0.55, 1.00]
])

plt.figure(figsize=(5, 4))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    square=True,
    cbar=True,
    fmt='.2f'
)
plt.title('Sample Correlation Matrix', fontsize=14)
plt.show()
```
